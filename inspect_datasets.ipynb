# Dataset Format Inspector

This notebook inspects the format and structure of parquet datasets:
- DailyDialog_train_limit1000.parquet
- NegotiationToM_Qwen-Qwen2.5-3B-Instruct_limit800.parquet

```python
import pandas as pd
import pyarrow.parquet as pq
import json
from pathlib import Path
```

## 1. DailyDialog Dataset

```python
# Load DailyDialog dataset
daily_dialog_path = "data/DailyDialog_train_limit1000.parquet"
df_daily = pd.read_parquet(daily_dialog_path)

print("=" * 80)
print("DAILYDIALOG DATASET")
print("=" * 80)
```

```python
# Basic information
print(f"\n📊 Shape: {df_daily.shape}")
print(f"   Rows: {df_daily.shape[0]:,}")
print(f"   Columns: {df_daily.shape[1]}")
```

```python
# Column information
print(f"\n📋 Columns: {list(df_daily.columns)}")
print(f"\n📝 Data Types:")
print(df_daily.dtypes)
```

```python
# Memory usage
print(f"\n💾 Memory Usage:")
print(df_daily.memory_usage(deep=True))
```

```python
# First few rows
print(f"\n🔍 First 3 Rows:")
df_daily.head(3)
```

```python
# Detailed view of first row
print(f"\n🔬 Detailed View of First Row:")
print("=" * 80)
for col in df_daily.columns:
    value = df_daily.iloc[0][col]
    print(f"\n{col}:")
    print(f"  Type: {type(value).__name__}")
    if isinstance(value, (list, dict)):
        print(f"  Value: {json.dumps(value, indent=4)}")
    else:
        print(f"  Value: {value}")
```

```python
# Check for null values
print(f"\n❌ Null Values:")
null_counts = df_daily.isnull().sum()
if null_counts.sum() > 0:
    print(null_counts[null_counts > 0])
else:
    print("  No null values found")
```

```python
# Statistical summary for numeric columns
numeric_cols = df_daily.select_dtypes(include=['int64', 'float64']).columns
if len(numeric_cols) > 0:
    print(f"\n📈 Statistical Summary (Numeric Columns):")
    print(df_daily[numeric_cols].describe())
```

```python
# Check unique values for each column
print(f"\n🔢 Unique Value Counts:")
for col in df_daily.columns:
    n_unique = df_daily[col].nunique()
    print(f"  {col}: {n_unique:,} unique values")
```

## 3. Schema Comparison

```python
print("\n" + "=" * 80)
print("SCHEMA COMPARISON")
print("=" * 80)

print(f"\n📋 Column Comparison:")
daily_cols = set(df_daily.columns)
negotiation_cols = set(df_negotiation.columns)

print(f"\n  Common columns: {daily_cols & negotiation_cols}")
print(f"  Only in DailyDialog: {daily_cols - negotiation_cols}")
print(f"  Only in NegotiationToM: {negotiation_cols - daily_cols}")
```

## 4. PyArrow Schema Details

```python
# Get PyArrow schema details for DailyDialog
print("\n" + "=" * 80)
print("PYARROW SCHEMA - DAILYDIALOG")
print("=" * 80)
parquet_file_daily = pq.ParquetFile(daily_dialog_path)
print(parquet_file_daily.schema)
```

```python
# Get PyArrow schema details for NegotiationToM
print("\n" + "=" * 80)
print("PYARROW SCHEMA - NEGOTIATIONTOM")
print("=" * 80)
parquet_file_negotiation = pq.ParquetFile(negotiation_path)
print(parquet_file_negotiation.schema)
```

## 6. Data Quality Checks

```python
# Check for duplicates in DailyDialog
print("\n" + "=" * 80)
print("DATA QUALITY - DAILYDIALOG")
print("=" * 80)
n_duplicates = df_daily.duplicated().sum()
print(f"\n🔍 Duplicate Rows: {n_duplicates}")
if n_duplicates > 0:
    print(f"   ({n_duplicates / len(df_daily) * 100:.2f}% of dataset)")
```

```python
# Check for duplicates in NegotiationToM
print("\n" + "=" * 80)
print("DATA QUALITY - NEGOTIATIONTOM")
print("=" * 80)
n_duplicates = df_negotiation.duplicated().sum()
print(f"\n🔍 Duplicate Rows: {n_duplicates}")
if n_duplicates > 0:
    print(f"   ({n_duplicates / len(df_negotiation) * 100:.2f}% of dataset)")
```

## 7. Export Summary to JSON

```python
# Create a summary dictionary
summary = {
    "DailyDialog": {
        "path": daily_dialog_path,
        "shape": df_daily.shape,
        "columns": list(df_daily.columns),
        "dtypes": {col: str(dtype) for col, dtype in df_daily.dtypes.items()},
        "null_counts": df_daily.isnull().sum().to_dict(),
        "unique_counts": {col: int(df_daily[col].nunique()) for col in df_daily.columns},
        "memory_usage_bytes": int(df_daily.memory_usage(deep=True).sum())
    },
    "NegotiationToM": {
        "path": negotiation_path,
        "shape": df_negotiation.shape,
        "columns": list(df_negotiation.columns),
        "dtypes": {col: str(dtype) for col, dtype in df_negotiation.dtypes.items()},
        "null_counts": df_negotiation.isnull().sum().to_dict(),
        "unique_counts": {col: int(df_negotiation[col].nunique()) for col in df_negotiation.columns},
        "memory_usage_bytes": int(df_negotiation.memory_usage(deep=True).sum())
    }
}

# Save summary to JSON
output_path = "dataset_inspection_summary.json"
with open(output_path, 'w') as f:
    json.dump(summary, f, indent=2)

print(f"\n✅ Summary exported to: {output_path}")
```
